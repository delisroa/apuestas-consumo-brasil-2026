# ¿Menos apuestas, más compras?
## Análisis exploratorio del comercio alimentario en Brasil (agosto–octubre de 2026)

**Pregunta:** ¿Se observa un cambio en la variación de ventas de supermercados después de las restricciones a las apuestas de cuota fija en Brasil?

El 25 de septiembre de 2026 se prohibieron nuevos depósitos en las plataformas y el 6 de octubre se dispuso el cierre de los sitios y aplicaciones, según el Ministerio de Hacienda de Brasil.

Para comparar períodos utilizo datos semanales de **variación interanual de ventas en valor (%)** publicados por NielsenIQ y ABRAS. Esto **no mide directamente el volumen de productos comprados** ni permite, por sí solo, atribuir los cambios a la prohibición.

**Fuentes:**
- [NielsenIQ / ABRAS — Monitoreo semanal, semana 42 de 2026 (gráfico de la página 2)](https://static.abras.com.br/pdf/monitoramento-semanal-niq/semana-4226.pdf)
- [ABRAS — Histórico de monitoreo semanal de 2026](https://www.abras.com.br/economia-e-pesquisa/monitoramento-semanal-nielseniq-e-abras/2026)
- [Ministerio de Hacienda de Brasil — Medida Provisional 1.394/2026 y cronograma](https://www.gov.br/fazenda/pt-br/composicao/orgaos/secretaria-de-premios-e-apostas/medida-provisoria-ndeg-1-394-2026-entenda-as-novas-regras-para-as-apostas-de-quota-fixa)

```python
import pandas as pd
import matplotlib.pyplot as plt

# Valores tomados del gráfico semanal de NielsenIQ/ABRAS (semana 42, pág. 2)
datos = {
    "semana": [
        "2026-08-02", "2026-08-09", "2026-08-16", "2026-08-23",
        "2026-08-30", "2026-09-06", "2026-09-13", "2026-09-20",
        "2026-09-27", "2026-10-04", "2026-10-11", "2026-10-18"
    ],
    "variacion_ventas": [-0.6, 5.1, 6.6, 6.9, -3.3, 8.7, 3.7, 7.9, 2.4, -0.3, 7.5, 7.2],
    "periodo": ["Antes"] * 8 + ["Transición"] * 2 + ["Después"] * 2
}

df = pd.DataFrame(datos)
df["semana"] = pd.to_datetime(df["semana"])
df
```

### Criterio de comparación

Dividí la muestra en **Antes** (8 semanas hasta el 20/09), **Transición** (2 semanas con fechas de cierre 27/09 y 04/10) y **Después** (2 semanas con fechas de cierre 11/10 y 18/10). Las semanas están identificadas por su fecha de cierre. Los nombres de los períodos son una **clasificación para este análisis** y no forman parte de la base original de ABRAS.

```python
# Estadísticas descriptivas de cada período
resumen = df.groupby("periodo", sort=False)["variacion_ventas"].agg(["count", "mean", "min", "max"]).round(2)
resumen
```

```python
# Evolución semanal y fechas de referencia
plt.figure(figsize=(11, 5))
plt.plot(df["semana"], df["variacion_ventas"], marker="o")
plt.axvline(pd.Timestamp("2026-09-25"), linestyle="--", color="red", label="Restricción de nuevos depósitos")
plt.axvline(pd.Timestamp("2026-10-06"), linestyle=":", color="green", label="Cierre de plataformas")
plt.axhline(0, color="gray", linewidth=1)
plt.title("Supermercados en Brasil: evolución semanal de ventas (2026)")
plt.xlabel("Semana (fecha de cierre)")
plt.ylabel("Variación interanual de ventas en valor (%)")
plt.xticks(df["semana"], df["semana"].dt.strftime("%d/%m"), rotation=45)
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Promedio de las variaciones semanales por período
resumen["mean"].plot(kind="bar", figsize=(8, 4))
plt.title("Variación interanual promedio según período")
plt.xlabel("Período")
plt.ylabel("Promedio de la variación interanual (%)")
plt.xticks(rotation=0)
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

### Resultados y conclusión preliminar

En estas 12 semanas, el promedio de las variaciones interanuales fue de **4,38 %** antes de las restricciones, **1,05 %** durante la transición y **7,35 %** después. La diferencia entre el promedio anterior y el posterior es de aproximadamente **2,97 puntos porcentuales**.

El período anterior ya contenía semanas con crecimientos del **8,7 %** y **7,9 %**, valores superiores a los observados en las dos semanas posteriores (**7,5 %** y **7,2 %**). Por lo tanto, la información disponible **no demuestra** que el aumento posterior sea excepcional ni que haya sido causado por las restricciones a las apuestas.

**Limitaciones:**
- Solo hay dos semanas clasificadas como posteriores.
- Se usa un indicador de **ventas en valor** (variación interanual), no cantidades compradas ni gasto individual de los hogares.
- No contamos con una serie comparable de gasto en apuestas.
- No controlamos inflación, estacionalidad, promociones ni diferencias regionales.
- El informe de ABRAS/NielsenIQ señala que **la apertura de nuevas tiendas explica gran parte del crecimiento acumulado** del sector en las últimas 52 semanas (87 %, página 5). Es otro factor que conviene considerar al interpretar los resultados.

**Balance:** el análisis identifica una coincidencia temporal y permite plantear preguntas nuevas, pero no prueba causalidad. Para avanzar sería necesario incorporar más semanas posteriores y datos adicionales sobre apuestas y consumo real.
